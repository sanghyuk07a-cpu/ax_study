```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('usa_housing.csv')
df.head()
```

```python
import pandas as pd
df = df[(df['YearBuilt'] > 2000) & (df['CrimeRate'] < 40)]

df.head()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['SquareFeet'], df['Price'], color='dodgerblue', alpha=0.5, edgecolor='w')
plt.title('SchoolRating vs Price')
plt.xlabel('SchoolRating')
plt.ylabel('Price')
plt.grid(True, linestyle='--')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
x = df[['SchoolRating']].values
y = df['Price'].values
```

```python
model = LinearRegression()
model.fit(x, y);
```

```python
print(model.coef_, model.intercept_)
```

```python
y_pred = model.predict(x)

y_pred
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='dodgerblue', alpha=0.5, edgecolor='w', label='Actual Data')
plt.plot(x, y_pred, color='crimson', linewidth=3, label='Regression Line')
plt.title('Linear Regression Prediction')
plt.xlabel('SchoolRating')
plt.ylabel('Price')
plt.legend()
plt.grid(True, linestyle='--')
plt.show()
```
