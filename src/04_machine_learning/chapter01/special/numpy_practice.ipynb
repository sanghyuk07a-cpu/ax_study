축(axis)

2차원 배열에서 연산의 방행을 설정
- axis: 0 -> 행기준에서 연산
- axis: 1 -> 열기준에서 연산

```python
mat = np.arange(1, 7).reshape(2, -1)
print(mat)

sum_axis0 = mat.sum(axis=0)
sum_axis1 = mat.sum(axis=1)

print("행 기준 합계:", sum_axis0)
print("열 기준 합계:", sum_axis1)
```

#### 1.4 행렬 전치(Transpose) 및 행렬곱(@)

- 객체.transpose() -> {행, 열} -> {열, 행} / 전치
- 객체.T
- 행렬 곱 연산(@)

```python
mat = np.array([
    [1, 2, 3],
    [4, 5, 6]
    ])

print(mat)
print("전치 행렬:\n", mat.transpose())
print("전치 행렬:\n", mat.T)
```

```python
mat1 = np.array([[10, 20], [30, 40]])
mat2 = np.array([[1,2], [3, 4]])

mat3 = mat1 @ mat2

mat3
```
